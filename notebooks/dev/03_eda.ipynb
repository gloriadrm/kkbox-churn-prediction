## CATEGORICAL EDA

### Statistical analysis

```python
def categorical_frequencies(df_cat):
    result = pd.concat(
        [
            df_cat[col]
            .value_counts(normalize=True, dropna=False)
            .mul(100)
            .rename_axis("value")
            .reset_index(name="frequency")
            .assign(variable=col)
            for col in df_cat.columns
        ],
        ignore_index=True
    )

    return (
        result[["variable", "value", "frequency"]]
        .sort_values(
            by=["variable", "frequency"],
            ascending=[True, True]
        )
        .reset_index(drop=True)
    )
```

```python
categorical_frequencies(cat)
```

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# 1. Comprobar renderizador
print("Renderer actual:", pio.renderers.default)

# 2. Construir DataFrame con tipos nativos
plot_df = (
    cat["registered_via"]
    .value_counts(normalize=True)
    .mul(100)
    .rename_axis("registered_via")
    .reset_index(name="porcentaje")
)

plot_df["registered_via"] = plot_df["registered_via"].astype(str)
plot_df["porcentaje"] = plot_df["porcentaje"].astype("float64")

print(plot_df)
print(plot_df.dtypes)

# 3. Gráfico
fig = px.bar(
    plot_df,
    x="porcentaje",
    y="registered_via",
    orientation="h",
    text="porcentaje",
    title="Distribución de registered_via"
)

fig.update_traces(texttemplate="%{x:.1f}%", textposition="auto")
fig.update_xaxes(range=[0, 55])
fig.show(renderer="browser")

```
