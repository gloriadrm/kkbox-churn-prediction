```python
%load_ext autoreload
%autoreload 2

import pandas as pd
from sklearn.metrics import confusion_matrix

from kkbox_churn.config import TARGET, VALIDATION_PATH
from kkbox_churn.data.load import load_dataset
from kkbox_churn.models.predict import load_model, predict
from kkbox_churn.models.evaluate import evaluate

val = load_dataset(VALIDATION_PATH)
model = load_model()

preds = predict(model, val)
metrics = evaluate(val[TARGET], preds["churn_proba"], preds["churn_pred"])

print(f"Umbral del modelo: {model.best_threshold_:.4f}")

display(pd.Series(metrics).round(4))
print(confusion_matrix(val[TARGET], preds["churn_pred"]))
```
