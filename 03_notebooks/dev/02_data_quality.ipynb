```python
# Categorical variable frequencies

def categorical_frequencies(df_cat):
    result = pd.concat(
        [
            df_cat[col]
            .value_counts(normalize=True, dropna=False)
            .mul(100)
            .rename_axis("value")
            .reset_index(name="frequency")
            .assign(variable=col)
            for col in df_cat.columns
        ],
        ignore_index=True
    )

    return (
        result[["variable", "value", "frequency"]]
        .sort_values(
            by=["variable", "frequency"],
            ascending=[True, True]
        )
        .reset_index(drop=True)
    )


# Frecuencias categóricas
cat_freq = categorical_frequencies(cat)

# Categorías con frecuencia inferior al 0,05 %
display(cat_freq[cat_freq["frequency"] < 0.05])
```

### Outliers

#### By median absolute deviation (MAD)

El Z-score supone una distribución más o menos normal. Las variables (importes, número de transacciones, días) son muy asimétricas, con muchos valores pequeños y una cola larga, con lo cual el Z-score funciona mal. 

Empleamos una alternativa robusta basada en la mediana y la MAD (desviación absoluta mediana), que no se dejan arrastrar por los extremos.

```python
def outliers_summary(df_num, k=5):
    """
    Resumen de outliers por variable usando mediana y MAD (robusto a extremos).
    
    'lower_lim' y 'upper_lim' son los límites a partir de los cuales un valor se considera outlier.
    Un lower_lim negativo en una variable que no puede ser negativa (días, importes, conteos) solo indica que por abajo no hay outliers posibles. No es un error.
    Los límites dependen de k. Si te salen demasiados outliers, sube k; si te salen muy pocos, bájalo. No son una verdad absoluta, solo un umbral de decsión.
    
    Parameters
    ----------
    df_num : DataFrame con variables numéricas
    k : nº de MADs para considerar un valor outlier (3 = estricto, 5 = moderado)
    """
    rows = []

    for col in df_num.columns:
        s = df_num[col].dropna()
        if s.empty:
            continue

        median = s.median()
        mad = 1.4826 * (s - median).abs().median()  # escalada para ser comparable a la sd

        if mad == 0:
            # Variable casi constante: usar percentiles como alternativa
            lower, upper = s.quantile(0.001), s.quantile(0.999)
        else:
            lower, upper = median - k * mad, median + k * mad

        mask = (s < lower) | (s > upper)

        rows.append({
            "variable": col,
            "n_outliers": mask.sum(),
            "pct_outliers": round(100 * mask.mean(), 2),
            "min": s.min(),
            "p1": s.quantile(0.01),
            "median": median,
            "p99": s.quantile(0.99),
            "max": s.max(),
            "lower_lim": lower,
            "upper_lim": upper,
        })

    return (pd.DataFrame(rows)
              .sort_values("pct_outliers", ascending=False)
              .reset_index(drop=True))
```
