# MODEL RE-TRAINING CODE

```python
from pathlib import Path
import numpy as np
import pandas as pd
import cloudpickle

from janitor import clean_names

from sklearn.preprocessing import OrdinalEncoder, KBinsDiscretizer
from sklearn.preprocessing import Binarizer, QuantileTransformer, StandardScaler

from xgboost import XGBClassifier

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline

from kkbox_churn.config import PROJECT_ROOT, SELECTED_FEATURES

# -----------------------------
# LOAD NEW RAW DATA
# -----------------------------
project_path = Path('/Users/gloriadelriomarquez/Documents/Career/Projects/ml-lab/client_churn')
df = pd.read_csv(project_path / "02_data" / "train" / "train.csv", index_col=0)

# -----------------------------
# BASIC CLEANING AND SELECTION
# -----------------------------
df = (
    clean_names(df)       # snake_case names
    .rename(columns = {'bd':'age'})
)

target = 'is_churn'
X = df[SELECTED_FEATURES].copy()   
y = df[target].copy()

# -----------------------------
# LOAD TRAINING PIPELINE
# -----------------------------
training_pipeline_path = PROJECT_ROOT / "04_models" / "training_pipeline.pickle"

with open(training_pipeline_path, mode="rb") as file:
    training_pipeline = cloudpickle.load(file)

# -----------------------------
# FIT EXECUTION PIPELINE ON NEW DATA
# -----------------------------
execution_pipeline = training_pipeline.fit(X, y)

# -----------------------------
# SAVE EXECUTION PIPELINE
# -----------------------------
execution_pipeline_path = PROJECT_ROOT / "04_models" / "execution_pipeline.pickle"

with open(execution_pipeline_path, mode="wb") as file:
    cloudpickle.dump(execution_pipeline, file)
```
